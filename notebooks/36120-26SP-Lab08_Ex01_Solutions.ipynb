# 3. Configure Supabase with DVC

**[3.1]** Login into your Supabase and create a project called `36120`

**[3.2]** Go to Storage and create a bucket called `dvc`

**[3.3]** Go to Storage > Settings > S3 Configuration. Generate a pair of credentials (Access Key ID and Secret Access Key) adn note them down.

**[3.4]** Note your Region and the direct S3-compatible Endpoint. Format: https://\<project-ref\>.storage.supabase.co/storage/v1/s3

**[3.5]** Add your Supabase bucket as a DVC remote

````python
! dvc remote add -d supabase-s3 s3://dvc
````

**[3.6]** Set the S3 endpoint URL to your Supabase storage URL

````python
! dvc remote modify supabase-s3 endpointurl https://<project-ref>.storage.supabase.co/storage/v1/s3
````

**[3.7]** Securely set your Supabase Access Key ID and Secret Access Key

````python
! dvc remote modify --local supabase-s3 access_key_id <your-access-key-id>
! dvc remote modify --local supabase-s3 secret_access_key <your-secret-access-key>
````
